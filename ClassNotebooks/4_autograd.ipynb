# Notebook 4: Autograd
In this notebook, we will have a look into how gradient calculation happens in pytorch

```python
import torch
```

```python
# Let's build a simple 1 layer network
x_in = torch.ones(5) # Input
y_out = torch.zeros(3) # Expected Output
w = torch.randn(5, 3, requires_grad=True) # Weights
b = torch.randn(3, requires_grad=True)  # Bias
x_out = torch.matmul(x_in, w)+b
loss = torch.nn.functional.binary_cross_entropy_with_logits(x_out, y_out)
```

```python
print(f"Gradient function for output = {x_out.grad_fn}")
print(f"Gradient function for loss = {loss.grad_fn}")
```

```python
# Computing Gradients
loss.backward()
print(w.grad)
print(b.grad)
```

```python
# Disabling Gradient Tracking
x_out = torch.matmul(x_in, w)+b
print(x_out.requires_grad)

with torch.no_grad():
    x_out = torch.matmul(x_in, w)+b
print(x_out.requires_grad)

x_out = torch.matmul(x_in, w)+b
x_out_det = x_out.detach()
print(x_out_det.requires_grad)
```
